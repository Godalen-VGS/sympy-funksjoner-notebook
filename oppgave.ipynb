## Funksjoner som virker både med tall og symboler

En vanlig Python-funksjon fungerer fint sammen med sympy. Gir vi den et tall, får vi et tall. Gir vi den symbolet `x`, får vi et uttrykk vi kan derivere.

```python
import sympy as sp

x = sp.symbols('x')

def f(x):
    return x**3 - 3*x

print(f(2))      # funksjonsverdien f(2)
print(f(x))      # uttrykket f(x)
```

## Derivasjon

`sp.diff(uttrykk, x)` deriverer med hensyn på $x$. Det svarer til `Derivert` i GeoGebra.

Topp- og bunnpunkter finner vi der den deriverte er null: vi løser $f'(x) = 0$.

```python
fd = sp.diff(f(x), x)
print("f'(x) =", fd)

kandidater = sp.solve(fd, x)
print("f'(x) = 0 for x =", kandidater)

for a in kandidater:
    print(f"Punktet ({a}, {f(a)})")
```

## Tegne grafen

Matplotlib trenger tall, ikke symboler. Den enkleste måten er å lage en liste med $x$-verdier med `np.linspace` og regne ut funksjonsverdiene. Har du bare et sympy-uttrykk, gjør `sp.lambdify(x, uttrykk)` det om til en vanlig funksjon som tåler numpy-lister.

```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.linspace(-2.5, 2.5, 200)
fd_tall = sp.lambdify(x, fd)          # gjør f'(x) om til en vanlig funksjon

plt.plot(xs, f(xs), label="f(x)")
plt.plot(xs, fd_tall(xs), "--", label="f'(x)")
plt.axhline(0, color="black", linewidth=0.8)
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
